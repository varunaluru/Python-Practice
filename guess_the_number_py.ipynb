```python
import random # Import the random module

# Generate a random integer between 1 and 50
secretNumber = random.randint(1,50)
# Initialize the attempt counter
attempt = 1
# Initialize the user's guess variable
x = 0

# Loop until the user guesses the correct number or runs out of attempts (6 attempts total)
while x != secretNumber and attempt < 6:
  # Get the user's guess as an integer input
  x = int(input("Guess the random number: "))

  # Check if the guess is too high
  if x > secretNumber:
    print("Too high")
    # If it's the last attempt and the guess is wrong, reveal the secret number
    if attempt == 5:
        print()
        print("Sorry that's not the right number.")
        print("The actual number was", secretNumber)
  # Check if the guess is too low
  elif x < secretNumber:
    print("Too low")
    # If it's the last attempt and the guess is wrong, reveal the secret number
    if attempt == 5:
        print()
        print("Sorry you didn't get it.")
        print("The actual number was", secretNumber)
  # If the guess is neither too high nor too low, it must be correct
  else:
    print("You got it!")

  print()

  # Increment the attempt counter
  attempt = attempt + 1

# Print a thank you message after the game ends
print("Thanks for playing!")
```
