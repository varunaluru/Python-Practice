```python
import random # Imports the random module

# Define the valid options for the game
options = ("rock", "paper", "scissors")
WINNING_SCORE = 5 # Set the score required to win the game

def get_player_name():
    """Gets the player's name from user input."""
    return input("Please enter your name: ")

def get_player_choice():
    """Prompts the player for their choice (Rock, Paper, Scissors) and validates it.
    Continues to prompt until a valid choice is entered.
    """
    while True:
        choice = input("Please choose Rock, Paper or Scissors: ").lower().strip() # Convert input to lowercase and remove whitespace
        if choice in options: # Check if the choice is one of the valid options
            return choice
        else:
            print("Invalid input. Please enter one of the options (Rock, Paper, Scissors).")

def determine_winner(player_choice, computer_choice):
    """Determines the winner of a single round based on player and computer choices.

    Args:
        player_choice (str): The player's choice.
        computer_choice (str): The computer's choice.

    Returns:
        str: "tie" if it's a draw, "player" if the player wins, "computer" if the computer wins.
    """
    if player_choice == computer_choice:
        return "tie"
    elif (
        (player_choice == "rock" and computer_choice == "scissors") or
        (player_choice == "paper" and computer_choice == "rock") or
        (player_choice == "scissors" and computer_choice == "paper")
    ):
        return "player"
    else:
        return "computer"

def display_scores(player_score, computer_score, player_name):
    """Displays the current scores of the player and the computer.


    """
    print(f"{player_name} has {player_score} point{'s' if player_score != 1 else ''}")
    print(f"The computer has {computer_score} point{'s' if computer_score != 1 else ''}")
    print()

def play_game():
    """Main function to play the Rock, Paper, Scissors game until one player reaches WINNING_SCORE."""
    player_name = get_player_name() # Get the player's name
    computer_score = 0 # Initialize computer's score
    player_score = 0   # Initialize player's score
    print(f"Welcome to the Rock, Paper and Scissors Game, {player_name}! ✊✋✌️ The first to get {WINNING_SCORE} points wins!")
    print()

    while True: # Game loop continues until a winner is determined
        # Check for win conditions after each round
        if computer_score == WINNING_SCORE:
            print(f"The computer got {WINNING_SCORE} points! The computer wins :(")
            break # Exit loop if computer wins
        if player_score == WINNING_SCORE:
            print(f"You got {WINNING_SCORE} points! You win!")
            break # Exit loop if player wins

        computers_choice = random.choice(options) # Computer makes a random choice
        players_choice = get_player_choice() # Get player's choice

        print()
        print(f"You chose {players_choice}")
        print(f"Computer chose {computers_choice}")

        winner = determine_winner(players_choice, computers_choice) # Determine round winner

        # Update scores and print round result
        if winner == "tie":
            print("It's a tie 🤝")
        elif winner == "player":
            player_score += 1
            print("You win 🏆")
        else: # winner == "computer"
            computer_score += 1
            print("You lose ❌")

        print()
        display_scores(player_score, computer_score, player_name) # Display current scores

# Start the game by calling the main play_game function
play_game()
```
