```python
import random #Imports the random function

options = ("rock","paper","scissors") #The options for the game
WINNING_SCORE = 5

def get_player_name():
    """Gets the player's name."""
    return input("Please enter your name: ")

def get_player_choice():
    """Prompts the player for their choice and validates it."""
    while True:
        choice = input("Please choose Rock, Paper or Scissors: ").lower().strip() # .strip() to remove leading/trailing whitespace
        if choice in options:
            return choice
        else:
            print("Invalid input. Please enter one of the options (Rock, Paper, Scissors).")

def determine_winner(player_choice, computer_choice):
    """Determines the winner of a single round."""
    if player_choice == computer_choice:
        return "tie"
    elif (
        (player_choice == "rock" and computer_choice == "scissors") or
        (player_choice == "paper" and computer_choice == "rock") or
        (player_choice == "scissors" and computer_choice == "paper")
    ):
        return "player"
    else:
        return "computer"

def display_scores(player_score, computer_score, player_name):
    """Displays the current scores."""
    print(f"{player_name} has {player_score} point{'s' if player_score != 1 else ''}")
    print(f"The computer has {computer_score} point{'s' if computer_score != 1 else ''}")
    print()

def play_game():
    """Main function to play the Rock, Paper, Scissors game."""
    player_name = get_player_name()
    computer_score = 0
    player_score = 0
    print(f"Welcome to the Rock, Paper and Scissors Game, {player_name}! The first to get {WINNING_SCORE} points wins!")
    print()

    while True:
        if computer_score == WINNING_SCORE:
            print("The computer got 5 points! The computer wins :(")
            break
        if player_score == WINNING_SCORE:
            print(f"You got {WINNING_SCORE} points! You win!")
            break

        computers_choice = random.choice(options)
        players_choice = get_player_choice()

        print()
        print(f"You chose {players_choice}")
        print(f"Computer chose {computers_choice}")

        winner = determine_winner(players_choice, computers_choice)

        if winner == "tie":
            print("It's a tie 🤝")
        elif winner == "player":
            player_score += 1
            print("You win 🏆")
        else:
            computer_score += 1
            print("You lose ❌")

        print()
        display_scores(player_score, computer_score, player_name)

# Start the game
play_game()
```
