```python
import random # Imports the random module

# --- Game Constants ---
NUM_ROUNDS = 3
DIE_MIN = 1
DIE_MAX = 6
CORRECT_GUESS_BONUS = 3
WINNING_SCORE_THRESHOLD = 14 # This threshold might be adjusted based on game balance

# --- Helper Functions ---
def get_valid_integer_input(prompt, min_val, max_val):
    """Gets and validates an integer input within a specified range."""
    while True:
        try:
            user_input_str = input(prompt)
            user_input = int(user_input_str)
            if min_val <= user_input <= max_val:
                return user_input
            else:
                print(f"Invalid input. Please enter a number between {min_val} and {max_val}.")
        except ValueError:
            print("Invalid input. Please enter a valid number.")

def get_yes_no_input(prompt):
    """Gets and validates a 'yes' or 'no' input."""
    while True:
        choice = input(prompt).lower()
        if choice in ['yes', 'no']:
            return choice
        else:
            print("Invalid choice. Please enter 'yes' or 'no'.")

def play_round(round_num, has_rerolled_this_game):
    """Plays a single round of the game and returns the round's score and reroll status."""
    print(f"\nRound {round_num} 🎯")

    player_guess = get_valid_integer_input(
        f"Guess the outcome of the die roll ({DIE_MIN}-{DIE_MAX}): ", DIE_MIN, DIE_MAX
    )

    print("Rolling the die 🎲...")
    initial_die_roll = random.randint(DIE_MIN, DIE_MAX) # Store initial roll for bonus check
    current_die_roll = initial_die_roll

    print(f"You rolled a {current_die_roll} 🎲")

    round_score = 0
    if player_guess == initial_die_roll:
        print(f"Your guess was correct! You get {CORRECT_GUESS_BONUS} bonus points! 🎉🌟")
        round_score += CORRECT_GUESS_BONUS

    # Handle re-roll choice
    if not has_rerolled_this_game: # Only ask if they haven't used their re-roll yet
        player_choice = get_yes_no_input("Do you want to re-roll? (yes/no): ")

        if player_choice == 'yes':
            print("Re-rolling the die... 🔄")
            current_die_roll = random.randint(DIE_MIN, DIE_MAX) # Re-rolls the die
            print(f"You re-rolled a {current_die_roll} 🎲") # Shows the new roll
            has_rerolled_this_game = True # Player has now used their re-roll ability
        else:
            print("No re-roll this round. 🤷‍♀️")
    else:
        print("You have already used your re-roll ability for this game. 🚫")

    round_score += current_die_roll # Add the final die roll of the round to the score
    return round_score, has_rerolled_this_game

def main_game():
    """Main function to run the dice guessing game."""
    has_rerolled = False # This keeps track if the player used this ability across all rounds
    total_score = 0 # This will keep track of the player's total score

    # This loop will run the game for the specified number of rounds
    for round_num in range(1, NUM_ROUNDS + 1):
        round_score, has_rerolled = play_round(round_num, has_rerolled)
        total_score += round_score

    print("\nGame Over! 🏁") # Announces the end of the game
    print(f"Your total score is: {total_score} points!") # Displays the player's final score
    if total_score > WINNING_SCORE_THRESHOLD:
        print("Congratulations! You won the game! 🏆🥳")
    else:
        print("Better luck next time! 😢❌")

# Run the game
if __name__ == "__main__":
    main_game()
```
