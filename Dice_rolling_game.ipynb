```python
import random # Imports the random module

has_rerolled = False # This variable keeps track if the player used this ability
total_score = 0 # This variable will keep track of the player's score

# This loop will run the game for 3 rounds
for round_num in range(3):
    print() # Prints an empty line for better readability
    print(f"Round {round_num + 1}") # Tells the player which round they are in
    print()
    print("Rolling the die...")
    die_roll = random.randint(1, 6) # Rolls a 6-sided die

    print(f"You rolled a {die_roll}") # Shows the player what they rolled

    # This loop handles the re-roll choice for the current round
    while True:
        # Asks the player if they want to re-roll and converts their answer to lowercase
        player_choice = input("Do you want to re-roll? (yes/no): ").lower()

        if player_choice == 'no':
            break # If the player doesn't want to re-roll, exit this loop
        elif player_choice == 'yes':
            # Check if the player has NOT used their re-roll ability yet
            if not has_rerolled:
                print() # Prints an empty line for better readability
                print("Re-rolling the die...")
                die_roll = random.randint(1, 6) # Re-rolls the die
                print(f"You re-rolled a {die_roll}") # Shows the new roll
                has_rerolled = True # Set this to True, they can't re-roll again
                break # Exit the re-roll loop after re-rolling
            else:
                # Tell the player if they've already used their re-roll
                print("You have already used your re-roll ability for this game.")
        else:
            # If the player types something other than 'yes' or 'no'
            print("Invalid choice. Please enter 'yes' or 'no'.")

    total_score += die_roll # Add the final die roll of the round to the total score

print() # Prints an empty line for better readability
print("Game Over!") # Announces the end of the game
print(f"Your total score is: {total_score}") # Displays the player's final score
```
