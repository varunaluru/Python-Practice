```python
import random

has_rerolled = False
total_score = 0

for round_num in range(3):
    print()
    print(f"Round {round_num + 1}")
    print()
    print("Rolling the die...")
    die_roll = random.randint(1, 6)

    print(f"You rolled a {die_roll}")

    while True:
        player_choice = input("Do you want to re-roll? (yes/no): ").lower()

        if player_choice == 'no':
            break
        elif player_choice == 'yes':
            if not has_rerolled:
                print()
                print("Re-rolling the die...")
                die_roll = random.randint(1, 6)
                print(f"You re-rolled a {die_roll}")
                has_rerolled = True
                break
            else:
                print("You have already used your re-roll ability for this game.")
        else:
            print("Invalid choice. Please enter 'yes' or 'no'.")

    total_score += die_roll

print()
print("Game Over!")
print(f"Your total score is: {total_score}")
```
