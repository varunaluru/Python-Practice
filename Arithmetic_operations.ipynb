```python
def add (x,y):
 return x+y
def sub (x,y):
 return x-y
def mul (x,y):
 return x*y
def div (x,y):
 return x/y

while True: # Main loop for the calculator
    while True:
        try:
            x = int(input("Enter the 1st number: "))
            if x < 0:
                print("Please enter a non-negative number.")
            else:
                break # Valid non-negative integer, exit loop
        except ValueError:
            print("Invalid input. Please enter numbers only.")

    while True:
        try:
            y = int(input("Enter the 2nd number: "))
            if y < 0:
                print("Please enter a non-negative number.")
            else:
                break # Valid non-negative integer, exit loop
        except ValueError:
            print("Invalid input. Please enter numbers only.")

    print("The addition of the 2 numbers you chose ")
    print(add(x,y))
    print("The subtraction of the 2 numbers you chose")
    print(sub(x,y))
    print(
"The multiplication of the 2 numbers you chose")
    print(mul(x,y))
    print("The division of the 2 numbers you chose")
    print(div(x,y))

    print(f"You entered: x={x}, y={y}")
    # Ask if the operations are over
    over = input("Is your operations over? (yes/no): ").strip().lower()

    if over == "yes":
        print(f"Have a nice day !\n")

        break # Exit the main calculator loop if operations are over
    elif over == "no":
        pass # Continue to the next iteration of the main loop
    else:
        print("Invalid input. Closing program.")
        break # Exit if input is invalid
```
