```python
# basic arithmetic functions
def add (x,y):
    return x+y
def sub (x,y):
    return x-y
def mul (x,y):
    return x*y
def div (x,y):
    # Handle division by zero to prevent errors
    if y == 0:
        return "Error: Cannot divide by zero."
    else:
        return x/y

# Main loop for the calculator to allow multiple operations
while True:
    # Loop for valid number input
    while True:
        try:
            x = int(input("Enter the 1st number: "))
            # Validation for non-negative numbers
            if x < 0:
                print("Please enter a non-negative number.")
            else:
                break # Valid non-negative integer, exit loop
        except ValueError:
            print("Invalid input. Please enter numbers only.")

    # Loop for valid second number input
    while True:
        try:
            y = int(input("Enter the 2nd number: "))
            # Validate for non-negative numbers
            if y < 0:
                print("Please enter a non-negative number.")
            else:
                break # Valid non-negative integer, exit loop
        except ValueError:
            print("Invalid input. Please enter numbers only.")

    # Perform and display the results of all operations
    print("The addition of the 2 numbers you chose ")
    print(add(x,y))
    print("The subtraction of the 2 numbers you chose")
    print(sub(x,y))
    print("The multiplication of the 2 numbers you chose")
    print(mul(x,y))
    print("The division of the 2 numbers you chose")
    print(div(x,y))

    print(f"You entered: x={x}, y={y}")

    # Ask if the operations are over, with input validation
    while True:
        over = input("Is your operations over? (yes/no): ").strip().lower()
        if over == "yes":
            print(f"Have a nice day !\n")
            break # Exit the loop for 'over' input
        elif over == "no":
            break
        else:
            print("Invalid input. Please answer 'yes' or 'no'.")

    # If the user wants to stop, break out of the loop
    if over == "yes":
        break
    # If 'over' is 'no', the main loop continues
```
