```python
import random

def play_hangman():
    # Display the visual banner for the game
    welcome_banner = """
      ====================
      🎮 H A N G M A N 🎮
      ===================="""
    print(welcome_banner)
    print(" 🎮 Welcome to the Hangman game!")

    # List of possible secret words
    words = [
        "volcano", "australia", "football", "crocodile", "number",
        "google", "apple", "coconut", "roblox", "suitcase",
        "dinosaur", "aura", "macbook", "paper", "photo",
        "portfolio", "kingdom", "throne", "elephant", "mystical",
        "interface", "programming"
    ]

    # ASCII art representing the gallows for different stages of wrong guesses
    hangman_art = [
        r"""
        +----+
             |
             |
             |
             |
             |
        =========
        """,
        r"""
        +----+
             |
         o   |
             |
             |
             |
        =======
        """,
        r"""
        +----+
             |
         o   |
         |   |
             |
             |
        =======
        """,
        r"""
        +----+
             |
         o   |
        /|   |
             |
             |
        =======
        """,
        r"""
        +----+
             |
         o   |
        /|\  |
             |
             |
        =======
        """,
        r"""
      +----+
             |
         o   |
        /|\  |
        /    |
             |
        =======
        """,
        r"""
        +----+
             |
         o   |
        /|\  |
        / \  |
             |
        =======
        """,
    ]

    # Randomly select a secret word and convert it to lowercase
    secret_word = random.choice(words).lower()

    # Initialize the hidden word with underscores for each letter
    hidden_word = ["_"] * len(secret_word)
    lives = 6

    # Use a set to store unique letters guessed by the player
    guessed_letters = set()

    print(f"The word has {len(secret_word)} letters:")
    print(" ".join(hidden_word))

    # Main game loop
    while lives > 0:
        # Show currently guessed letters sorted alphabetically
        print(f"\nGuessed letters: {', '.join(sorted(guessed_letters)) if guessed_letters else 'None'}")
        guess = input(" 📣 Enter a letter: ").lower().strip()

        # Validate input: must be a single alphabetic letter
        if not guess.isalpha() or len(guess) != 1:
            print(" ⚠️ Invalid input! Please enter a single letter.")
            continue

        # Check if the letter has already been guessed
        if guess in guessed_letters:
            print(f" 🤔 You've already guessed '{guess}'. Try a different letter.")
            continue

        # Record the new guess
        guessed_letters.add(guess)

        # Check if the guess is in the secret word
        if guess in secret_word:
            print(f" ✅ Good guess! '{guess}' is in the word.")
            # Uncover the guessed letter in all its matching positions
            for position, letter in enumerate(secret_word):
                if letter == guess:
                    hidden_word[position] = guess
        else:
            # Deduct a life and show the corresponding hangman art stage
            lives -= 1
            print(f" ❌ Wrong guess! You have {lives} ❤️ lives left.")
            print(hangman_art[6 - lives])

        # Show the updated layout of the hidden word
        print(" ".join(hidden_word))

        # Check for victory condition (no underscores remaining)
        if "_" not in hidden_word:
            print(" 🎊 Congratulations, you guessed the word! 🏆 ")
            print(f" The word was 🔎 {secret_word.upper()}.")
            break
    else:
        # Triggers if the player runs out of lives
        print(f" 💀 Game over! The word was 🔎 {secret_word.upper()}.")
        print(hangman_art[6])

# Runs the game
play_hangman()
```
