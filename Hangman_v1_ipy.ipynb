```python
nimport random # imports the random function

# Assign the multiline string to a variable first to avoid potential parsing issues
welcome_banner = """
  ====================
  🎮 H A N G M A N 🎮
  ===================="""
print(welcome_banner) # prints the beginning of the game
print(" 🎮 Welcome to the Hangman game") # Also prints the beginning of the game
words = [
    "volcano",
    "australia",
    "football",
    "crocodile",
    "number",
    "google",
    "apple",
    "coconut",
    "roblox",
    "suitcase",
    "dinosaur",
    "aura",
    "macbook",
    "paper",
    "photo",
    "portfolio",
    "kingdom",
    "throne",
    "elephant",
    "mystical",
    "interface",
    "programming"

] # The list of words the player needs to guess from
hangman_art = [
    r"""
    +----+
         |
         |
         |
         |
         |
    =========
    """,
    r"""
    +----+
         |
     o   |
         |
         |
         |
    =======
    """,
    r"""
    +----+
         |
     o   |
     |   |
         |
         |
    =======
    """,
    r"""
    +----+
         |
     o   |
    /|   |
         |
         |
    =======
    """,
    r"""
    +----+
         |
     o   |
    /|\  |
         |
         |
    =======
    """,
    r"""
  +----+
         |
     o   |
    /|\  |
    /    |
         |
    =======
    """,
    r"""
    +----+
         |
     o   |
    /|\  |
    / \  |
         |
    =======
    """,
] # All of the hangman art when you get things wrong or correct.

secret_word = random.choice(words).lower() # Computer picks a random word out of the words, convert to lowercase

hidden_word = [] # The word computer chose is hidden

for letter in secret_word:
    hidden_word.append("_") # For the amount of letter in the secret word computer adds the _

lives = 6 # The amount of lives the player has
guessed_letters = [] # New: Keep track of guessed letters

print(f"The word has {len(secret_word)} letters:") # Indicate word length
for item in hidden_word:
    print(item, end=" ") # Used to show the amount of letters the word has

print() # prints a blank line

while lives > 0: # Shows if the game is still continuing the enter a letter will still be there
    print(f"\nGuessed letters: {', '.join(sorted(guessed_letters))}") # New: Display guessed letters
    guess = input(" 🔤 Enter a letter: ").lower() # What it shows when the player needs to type a letter, convert to lowercase

    # New: Input validation
    if not guess.isalpha():
        print(" ⚠️ Invalid input! Please enter a letter.")
        continue
    if len(guess) != 1:
        print(" ⚠️ Invalid input! Please enter only one letter.")
        continue
    if guess in guessed_letters:
        print(f" 🤔 You've already guessed '{guess}'. Try a different letter.")
        continue

    guessed_letters.append(guess) # New: Add guess to guessed letters list

    guess_correct = False

    for position, letter in enumerate(secret_word): # Adds the letter if guess it correct

        if letter == guess:
            hidden_word[position] = letter
            guess_correct = True

    if not guess_correct:
       lives -= 1 # If the player guesses wrong they lose a life
       print(f"  ❌ Wrong guess! You have {lives} ❤️ lives left.")
       if lives >= 0: # Ensure index is not out of bounds
           print(hangman_art[6 - lives])
    else:
        print(f" ✅ Good guess! '{guess}' is in the word.") # New: Inform about correct guess

    for item in hidden_word:
        print(item, end=" ") # Puts the space between the _

    print()

    if "_" not in hidden_word:
        print(" 🎊 Congratulations, you guessed the word! 🏆 ") # If the player guesses the word right
        print(f" The word was 🔎 {secret_word.upper()}.") # The word gets revealed in uppercase
        break

    if lives == 0:
        print(f" 💀 Game over! The word was 🔎 {secret_word.upper()}.") # When the player has 0 lives, reveal in uppercase
        print(hangman_art[6]) # Display final hangman art
        break
```
