```python
import random # imports the random function
print("""
  ====================
  🎮 H A N G M A N 🎮
  ====================""") # prints the beginning of the game
print(" 🎮 Welcome to the Hangman game") # Also prints the beginning of the game
words = [
    "volcano",
    "australia",
    "football",
    "crocodile",
    "number",
    "google",
    "apple",
    "coconut",
    "roblox",
    "suitcase",
    "dinosaur",
    "aura",
    "macbook",
    "paper",
    "photo",
    "portfolio",
    "kingdom",
    "throne",
    "elephant",
    "mystical",
    "interface",
    "programming"

] # The list of words the player needs to guess from
hangman_art = [
    """
    +----+
         |
         |
         |
         |
         |
    =========
    """,
    """
    +----+
         |
     o   |
         |
         |
         |
    =======
    """,
    """
    +----+
         |
     o   |
     |   |
         |
         |
    =======
    """,
    """
    +----+
         |
     o   |
    /|   |
         |
         |
    =======
    """,
    """
    +----+
         |
     o   |
    /|\\  |
         |
         |
    =======
    """,
    """
  +----+
         |
     o   |
    /|\\  |
    /    |
         |
    =======
    """,
    """
    +----+
         |
     o   |
    /|\\  |
    / \\  |
         |
    =======
    """,
] # All of the hangman art when you get things wrong or correct.

secret_word = random.choice(words) # Computer picks a random word out of the words

hidden_word = [] # The word computer chose is hidden

for letter in secret_word:
    hidden_word.append("_") # For the amount of letter in the secret word computer adds the _

lives = 6 # The amount of lives the player has

for item in hidden_word:
    print(item, end=" ") # Used to show the amount of letters the word has

print() # prints a blank line

while lives > 0: # Shows if the game is still continuing the enter a letter will still be there

    guess = input(" 🔤 Enter a letter: ") # What it shows when the player needs to type a letter

    guess_correct = False

    for position, letter in enumerate(secret_word): # Adds the letter if guess it correctb

        if letter == guess:
            hidden_word[position] = letter
            guess_correct = True

    if not guess_correct:
       lives -= 1 # If the player guesses wrong they lose a life
       print(f"  ❌ Wrong guess! You have {lives} ❤️ lives left.")
       print(hangman_art[6 - lives])

    for item in hidden_word:
        print(item, end=" ") # Puts the space between the _

    print()

    if "_" not in hidden_word:
        print(" 🎊 Congratulations, you guessed the word! 🏆 ") # If the player guesses the word right
        print(f" The word was 🔎 {secret_word}.") # The word gets revealed
        break

    if lives == 0:
        print(f" 💀 Game over! The word was 🔎 {secret_word}.") # When the player has 0 lives
        break
```
